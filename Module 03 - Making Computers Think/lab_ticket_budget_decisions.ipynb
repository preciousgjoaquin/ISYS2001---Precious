## Your final prompt

> Paste your complete, finished prompt below (double-click to edit).

```
Create a simple Python budget decision tool called "Smart Purchase Advisor".

The program should ask the user for:

1.   Their monthly income
2.   Their current available savings
3. The price of the item they want to buy

The program should then calculate:
- what percentage of their monthly income the purchase price represents
- how much savings would remain after buying the item

Use these decision rules exactly:

- If the item costs more than the user's available savings, print: "You cannot afforf this purchase with your current savings."

- Otherwise, if the item costs more than 50% of the user's monthly income, print: "This is a very expensinve purchase. Consider waiting or saving more."

- Otherwise, if the item costs between 20% and 50% of the user's monthly income, print: "You may be able to afford thism but think carefully before buying."

- Otherwise, if the item costs less than 20% of the user's monthly income, print: "This purchase looks affordable based on your income and savings."

Also print:
- the purchase price as a percnetage of monthly income
- the mount of savings remaining after the purchase

Requirements:
- Use Python only
- Use input(), float(), if/elif/else, calculationsm and print()
- Do not use functions, loops, classes, external libraries, or advanced Python features
- Format money values to 2 decimal places
- Format percentages to 1 decimal place
- The program must run from top to bottom without needing any follow-up questions
- Include short comments in the code explaining each section




```

````python
# Paste the code your prompt produced here, then run it to show it works.

# --- Smart Purchase Advisor ---

# Get user input for monthly income, current savings, and item price
monthly_income = float(input("Enter your monthly income: "))
available_savings = float(input("Enter your current available savings: "))
item_price = float(input("Enter the price of the item you want to buy: "))

# Calculate the percentage of monthly income the purchase represents
# and the savings remaining after purchase
percentage_of_income = (item_price / monthly_income) * 100
savings_after_purchase = available_savings - item_price

# Apply decision rules
# If the item costs more than current savings
if item_price > available_savings:
    print("You cannot afford this purchase with your current savings.")
# Otherwise, if the item costs more than 50% of monthly income
elif percentage_of_income > 50.0:
    print("This is a very expensive purchase. Consider waiting or saving more.")
# Otherwise, if the item costs between 20% and 50% of monthly income
elif 20.0 <= percentage_of_income <= 50.0:
    print("You may be able to afford this, but think carefully before buying.")
# Otherwise, if the item costs less than 20% of monthly income
else:
    print("This purchase looks affordable based on your income and savings.")

# Print calculated results
# Format purchase price as a percentage of monthly income to 1 decimal place
print(f"The purchase price is {percentage_of_income:.1f}% of your monthly income.")
# Format savings remaining after purchase to 2 decimal places
print(f"Your savings remaining after the purchase would be ${savings_after_purchase:.2f}.")
````
