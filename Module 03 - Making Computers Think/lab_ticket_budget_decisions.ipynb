## Your final prompt

> Paste your complete, finished prompt below (double-click to edit).

```
Create a beginner-friendly Python program called "Purchase Advisor."

1. Ask the user to enter:
  - Their monthly budget (float)
  - The price of the itemy they want to buy (float)
  - How much they have already spent this month (float)

2. Calculate
  - Remaining budget = monthly budget - amount already spent
  - Remaining budget after purchase = remmaining budget - item price

3. Use these decision rules:
  - If the item price is greater than the remaining budget, print: "Recommendation: Do not buy it. You do not have enough money left in your budget."
  - Else if the item price is more than 50% of the remaining budget, print: "Recommendation: Think carefully before buying. This is a large expense."
  - Otherwise, print: "Recommendation: This purchase fits within your budget."

4. After the recommendation, also print:
  - Remaning budget before purchase
  - Remaning budget after purchase

5. Use only Python concepts:
  - input()
  - float()
  - variables
  - if, elif, else
  - arithmetic operators
  - print()
  - f-strings

6. Include comments explaining each section of the code.

The program should run from start to finish without requiring any additional changes or user clarification.


```

````python
# Purchase Advisor

# Ask the user for information
monthly_budget = float(input("Enter your monthly budget: $"))
item_price = float(input("Enter the price of the item: $"))
amount_spent = float(input("Enter how much you have already spent this month: $"))

# Calculate remaining budget
remaining_budget = monthly_budget - amount_spent
remaining_after_purchase = remaining_budget - item_price

# Make a recommendation
if item_price > remaining_budget:
    print("Recommendation: Do not buy it. You do not have enough money left in your budget.")
elif item_price > (remaining_budget * 0.5):
    print("Recommendation: Think carefully before buying. This is a large expense.")
else:
    print("Recommendation: This purchase fits within your budget.")

# Display the results
print(f"Remaining budget before purchase: ${remaining_budget:.2f}")
print(f"Remaining budget after purchase: ${remaining_after_purchase:.2f}")
````
