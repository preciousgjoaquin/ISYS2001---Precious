### Paste and Test Your AI-Generated Budget Tracker

```python
# AI-GENERATED BUDGET TRACKER CODE
# Ask for the user's monthly income
income = float(input("Enter your monthly income: "))

# Ask for up to 3 expenses
expense1 = float(input("Enter your first expense: "))
expense2 = float(input("Enter your second expense: "))
expense3 = float(input("Enter your third expense: "))

# Calculate total expenses
total_expenses = expense1 + expense2 + expense3

# Calculate remaining budget
remaining_budget = income - total_expenses

# Display the results
print("\nMonthly Income:", income)
print("Total Expenses:", total_expenses)
print("Remaining Budget:", remaining_budget)

# Warn the user if the budget is negative
if remaining_budget < 0:
    print("Warning: Your budget is negative!")
else:
    print("Good job! Your budget is within your income.")
```

## Part 4: Enhancement Challenge

Pick ONE feature to add with AI's help:
- 🏷️ Expense categories (food, transport, entertainment)
- 📊 Percentage of income spent
- ✅ Input validation (no negative numbers)
- 📝 A detailed budget summary with advice


```python
# Document your enhancement choice and the prompt you used
enhancement_choice = "I chose to add: Percentage of income spent"
enhancement_prompt = """
Help me add a feature to my Python budget tracker that calculates and displays the percentage of monthly income spent.
"""

print(f"🚀 Enhancement: 📊 Percentage of income spent")
```

```python
# ENHANCED BUDGET TRACKER
# [Paste your enhanced code here]
```

```python
# ENHANCED BUDGET TRACKER

# Take the user's monthly income
try:
    monthly_income = float(input("Enter your monthly income: "))
except ValueError:
    print("Invalid input for income. Please enter a number.")
    monthly_income = 0.0

expenses = []
# Takes up to 3 expenses
for i in range(3):
    expense_name = input(f"Enter name for expense {i+1} (or leave blank to skip): ")
    if not expense_name:
        break
    try:
        expense_amount = float(input(f"Enter amount for {expense_name}: "))
        expenses.append({'name': expense_name, 'amount': expense_amount})
    except ValueError:
        print("Invalid input for expense amount. Please enter a number.")

# Calculate what's left over
total_expenses = sum(expense['amount'] for expense in expenses)
remaining_budget = monthly_income - total_expenses

print(f"\n--- Budget Summary ---")
print(f"Monthly Income: ${monthly_income:.2f}")
print("Expenses:")
if expenses:
    for exp in expenses:
        print(f"  - {exp['name']}: ${exp['amount']:.2f}")
else:
    print("  (No expenses entered)")
print(f"Total Expenses: ${total_expenses:.2f}")
print(f"Remaining Budget: ${remaining_budget:.2f}")

# Calculate and display the percentage of income spent (Enhancement)
if monthly_income > 0:
    percentage_spent = (total_expenses / monthly_income) * 100
    print(f"Percentage of income spent: {percentage_spent:.2f}%")
else:
    print("Percentage of income spent: Cannot calculate (Monthly income is zero).")

# Warns the user if the budget goes negative
if remaining_budget < 0:
    print("\n⚠️ Warning: Your budget is negative! You've spent more than your income.")
elif remaining_budget == 0:
    print("\nHeads up: Your budget is balanced, with nothing left over.")
else:
    print("\nGood job! You have money left in your budget.")
```
